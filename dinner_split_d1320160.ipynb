## 作業說明

本作業使用 Python 變數、字串方法、型別轉換、數字運算、布林比較與 f-string，整理聚餐資料並計算服務費、折價後總額及每人分攤費用。

計算規則：
- 服務費以原始餐費的 10% 計算。
- 應付總額＝原始餐費＋服務費－折價券。
- 每人費用＝應付總額÷分攤人數。
- 每人費用大於 300 元時，判斷為超過預算。

## 一、測試一：原始資料

使用原始資料：4 人、折價券 100 元。

```python
# 1. 原始資料：保留文字資料，另外建立整理後的變數。
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# 2. 整理聚餐名稱。
clean_event_name = event_name.strip().lower()

# 3. 清理餐費文字，再轉換餐費與人數的型別。
clean_meal_text = meal_amount.replace("NT$", "").replace(",", "").replace(" ", "")
meal_value = float(clean_meal_text)
people_count = int(people)

# 4. 先按原始餐費計算服務費，再扣除折價券。
service_fee = meal_value * service_rate
total_amount = meal_value + service_fee - discount
amount_per_person = total_amount / people_count

# 5. 比較結果直接保存為 bool。等於預算時不算超過。
is_over_budget = amount_per_person > budget_per_person

# 6. 輸出時才將金額格式化為兩位小數。
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{meal_value:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people_count} 人")
print(f"每人費用：{amount_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```

## 三、測試三：折價券改成 220 元

從原始資料重新開始，人數恢復為 `"4"`，只將 `discount` 改成 220。重新執行完整程式。

```python
# 1. 原始資料：保留文字資料，另外建立整理後的變數。
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 220
budget_per_person = 300

# 2. 整理聚餐名稱。
clean_event_name = event_name.strip().lower()

# 3. 清理餐費文字，再轉換餐費與人數的型別。
clean_meal_text = meal_amount.replace("NT$", "").replace(",", "").replace(" ", "")
meal_value = float(clean_meal_text)
people_count = int(people)

# 4. 先按原始餐費計算服務費，再扣除折價券。
service_fee = meal_value * service_rate
total_amount = meal_value + service_fee - discount
amount_per_person = total_amount / people_count

# 5. 比較結果直接保存為 bool。等於預算時不算超過。
is_over_budget = amount_per_person > budget_per_person

# 6. 輸出時才將金額格式化為兩位小數。
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{meal_value:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people_count} 人")
print(f"每人費用：{amount_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```

## 四、程式逐步說明

### 1. 名稱整理
`event_name.strip().lower()` 先去掉頭尾空白，再轉成小寫，得到 `weekend dinner`。字串方法會回傳新字串，因此用 `clean_event_name` 保存，原始資料不變。

### 2. 餐費清理與型別轉換
`replace()` 移除 `NT$`、逗號及空白，得到文字 `1200`。再用 `float()` 轉成數值 `1200.0`。人數則使用 `int(people)` 轉成整數。

### 3. 計算順序
服務費以原始餐費計算，所以三組測試的服務費都是 1,200 × 0.10 = 120 元。接著加上服務費、扣除折價券，再除以人數。

### 4. 布林比較
`amount_per_person > budget_per_person` 會直接產生 `True` 或 `False`。因為題目問的是「是否超過」，所以使用 `>`；等於預算時結果為 `False`。

### 5. 輸出格式
f-string 中的 `:.2f` 讓金額顯示到小數點後兩位，例如 305.0 顯示為 305.00。格式化只影響顯示，不會改變原本數值。

## 六、兩題問答

### 問題一：為什麼 `"4"` 要先轉成數字？

`"4"` 是字串（`str`），`4` 才是整數（`int`）。本題需要使用人數作為除數，計算每人分攤費用。Python 無法直接將餐費數值除以字串，因此先用 `int(people)` 得到整數 `4`，再參與計算。

例如 `"4" + "1"` 會得到 `"41"`，而 `4 + 1` 才會得到 `5`。

### 問題二：如果先扣折價券，再計算服務費？

第一組先扣折價券：

1,200 − 100 = 1,100 元

再以 1,100 元計算 10% 服務費：

服務費 = 1,100 × 0.10 = 110 元

應付總額 = 1,100 + 110 = **1,210 元**

每人費用 = 1,210 ÷ 4 = **302.50 元**

題目規定的計算則是：

服務費 = 1,200 × 0.10 = 120 元

應付總額 = 1,200 + 120 − 100 = **1,220 元**

每人費用 = 1,220 ÷ 4 = **305.00 元**

兩種算法會差 10 元，是因為計算服務費時使用的金額不一樣。因此這題要按照題目規定，先用原始餐費計算服務費，再扣除折價券。
